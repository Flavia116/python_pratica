```python
# Calculadora de proteína por peso corporal

# Entrada
nome = input("Informe seu nome: ").upper()
peso = float(input("Digite seu peso: "))

# Processamento
proteina = peso * 2

# Saída
print(f"Olá, {nome}.")
print(f"A quantidade de proteína recomendada para seu peso é: {proteina:.1f} g")
```
