```python
# Calculadora de proteína, carboidrato e água para emagrecimento

# Entrada
nome = input("Informe seu nome: ").upper()
peso = float(input("Digite seu peso (kg): "))
refeicoes = int(input("Quantas refeições você faz por dia? "))

# Processamento
proteina_total = peso * 2
carboidrato_total = peso * 3
agua = peso * 33

proteina_refeicao = proteina_total / refeicoes
carboidrato_refeicao = carboidrato_total / refeicoes

# Saída
print(f"\nOlá, {nome}!")
print(f"\nNecessidades diárias:")
print(f"Proteína: {proteina_total:.1f} g")
print(f"Carboidrato: {carboidrato_total:.1f} g")
print(f"Água: {agua:.0f} mL")

print(f"\nDividindo em {refeicoes} refeições:")
print(f"Proteína por refeição: {proteina_refeicao:.1f} g")
print(f"Carboidrato por refeição: {carboidrato_refeicao:.1f} g")
```
